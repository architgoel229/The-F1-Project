```python
def get_mean_lap_time(df):
    if df.empty or "LapTime" not in df.columns:
        return np.nan

    values = df["LapTime"].dropna()

    if values.empty:
        return np.nan

    return values.mean()


def get_median_lap_time(df):
    if df.empty or "LapTime" not in df.columns:
        return np.nan

    values = df["LapTime"].dropna()

    if values.empty:
        return np.nan

    return values.median()
```

```python
def create_data_for_round(data):

    rows=[]
    laps = data["LapNumber"].dropna().unique()

    for lap in laps:

        if lap < 5:
            continue

        if lap > data["LapNumber"].max() - 5:
            break

        for driver in data[data["LapNumber"] == lap]["Driver"].unique():


            current = data[
                (data["LapNumber"] == lap) &
                (data["Driver"] == driver)
            ]

            if current.empty:
                continue

            current = current.iloc[0]

            Driver = driver
            Position = current["Position"]
            Time = current["Time"]
            Team = current["Team"]
            Compound = current["Compound"]
            TyreLife = current["TyreLife"]
            LapTime = current["LapTime"]
            current_lap_start = current["LapStartTime"]

            if pd.isna(Position):
                continue

            Position = int(Position)


            p = data[
                (data["LapNumber"] <= lap) &
                (data["LapNumber"] > lap - 5) &
                (data["Driver"] == driver)
            ]

            if p["LapTime"].notna().sum() < 4:
                continue


            p_minus1_driver = data[
                (data["LapNumber"] == lap) &
                (data["Position"] == Position - 1)
            ]["Driver"]

            p_minus2_driver = data[
                (data["LapNumber"] == lap) &
                (data["Position"] == Position - 2)
            ]["Driver"]

            p_minus3_driver = data[
                (data["LapNumber"] == lap) &
                (data["Position"] == Position - 3)
            ]["Driver"]

            p_plus1_driver = data[
                (data["LapNumber"] == lap) &
                (data["Position"] == Position + 1)
            ]["Driver"]

            p_plus2_driver = data[
                (data["LapNumber"] == lap) &
                (data["Position"] == Position + 2)
            ]["Driver"]

            p_plus3_driver = data[
                (data["LapNumber"] == lap) &
                (data["Position"] == Position + 3)
            ]["Driver"]

            p_minus1_driver = (
                p_minus1_driver.iloc[0]
                if not p_minus1_driver.empty
                else None
            )

            p_minus2_driver = (
                p_minus2_driver.iloc[0]
                if not p_minus2_driver.empty
                else None
            )

            p_minus3_driver = (
                p_minus3_driver.iloc[0]
                if not p_minus3_driver.empty
                else None
            )

            p_plus1_driver = (
                p_plus1_driver.iloc[0]
                if not p_plus1_driver.empty
                else None
            )

            p_plus2_driver = (
                p_plus2_driver.iloc[0]
                if not p_plus2_driver.empty
                else None
            )

            p_plus3_driver = (
                p_plus3_driver.iloc[0]
                if not p_plus3_driver.empty
                else None
            )


            p_minus1 = data[
                (data["Driver"] == p_minus1_driver) &
                (data["LapNumber"] <= lap) &
                (data["LapNumber"] > lap - 5)
            ].copy() if p_minus1_driver is not None else data.iloc[0:0].copy()

            p_minus2 = data[
                (data["Driver"] == p_minus2_driver) &
                (data["LapNumber"] <= lap) &
                (data["LapNumber"] > lap - 5)
            ].copy() if p_minus2_driver is not None else data.iloc[0:0].copy()

            p_minus3 = data[
                (data["Driver"] == p_minus3_driver) &
                (data["LapNumber"] <= lap) &
                (data["LapNumber"] > lap - 5)
            ].copy() if p_minus3_driver is not None else data.iloc[0:0].copy()

            p_plus1 = data[
                (data["Driver"] == p_plus1_driver) &
                (data["LapNumber"] <= lap) &
                (data["LapNumber"] > lap - 5)
            ].copy() if p_plus1_driver is not None else data.iloc[0:0].copy()

            p_plus2 = data[
                (data["Driver"] == p_plus2_driver) &
                (data["LapNumber"] <= lap) &
                (data["LapNumber"] > lap - 5)
            ].copy() if p_plus2_driver is not None else data.iloc[0:0].copy()

            p_plus3 = data[
                (data["Driver"] == p_plus3_driver) &
                (data["LapNumber"] <= lap) &
                (data["LapNumber"] > lap - 5)
            ].copy() if p_plus3_driver is not None else data.iloc[0:0].copy()


            p_minus1_mean_LapTime = get_mean_lap_time(p_minus1)
            p_minus2_mean_LapTime = get_mean_lap_time(p_minus2)
            p_minus3_mean_LapTime = get_mean_lap_time(p_minus3)

            p_plus1_mean_LapTime = get_mean_lap_time(p_plus1)
            p_plus2_mean_LapTime = get_mean_lap_time(p_plus2)
            p_plus3_mean_LapTime = get_mean_lap_time(p_plus3)


            p_minus1_median_LapTime = get_median_lap_time(p_minus1)
            p_minus2_median_LapTime = get_median_lap_time(p_minus2)
            p_minus3_median_LapTime = get_median_lap_time(p_minus3)

            p_plus1_median_LapTime = get_median_lap_time(p_plus1)
            p_plus2_median_LapTime = get_median_lap_time(p_plus2)
            p_plus3_median_LapTime = get_median_lap_time(p_plus3)


            p_3_minus1 = p_minus1[
                p_minus1["LapNumber"] > lap - 3
            ]

            p_3_minus2 = p_minus2[
                p_minus2["LapNumber"] > lap - 3
            ]

            p_3_minus3 = p_minus3[
                p_minus3["LapNumber"] > lap - 3
            ]

            p_3_plus1 = p_plus1[
                p_plus1["LapNumber"] > lap - 3
            ]

            p_3_plus2 = p_plus2[
                p_plus2["LapNumber"] > lap - 3
            ]

            p_3_plus3 = p_plus3[
                p_plus3["LapNumber"] > lap - 3
            ]
            p_3_minus1_mean_LapTime = get_mean_lap_time(p_3_minus1)
            p_3_minus2_mean_LapTime = get_mean_lap_time(p_3_minus2)
            p_3_minus3_mean_LapTime = get_mean_lap_time(p_3_minus3)

            p_3_plus1_mean_LapTime = get_mean_lap_time(p_3_plus1)
            p_3_plus2_mean_LapTime = get_mean_lap_time(p_3_plus2)
            p_3_plus3_mean_LapTime = get_mean_lap_time(p_3_plus3)


            p_3_minus1_median_LapTime = get_median_lap_time(p_3_minus1)
            p_3_minus2_median_LapTime = get_median_lap_time(p_3_minus2)
            p_3_minus3_median_LapTime = get_median_lap_time(p_3_minus3)

            p_3_plus1_median_LapTime = get_median_lap_time(p_3_plus1)
            p_3_plus2_median_LapTime = get_median_lap_time(p_3_plus2)
            p_3_plus3_median_LapTime = get_median_lap_time(p_3_plus3)
            def get_last_lap_time(df):

                if df.empty:
                    return np.nan

                df = df.dropna(subset=["LapTime"])

                if df.empty:
                    return np.nan

                return df.sort_values("LapNumber")["LapTime"].iloc[-1]

            p_minus1_LapTime = get_last_lap_time(p_minus1)
            p_minus2_LapTime = get_last_lap_time(p_minus2)
            p_minus3_LapTime = get_last_lap_time(p_minus3)

            p_plus1_LapTime = get_last_lap_time(p_plus1)
            p_plus2_LapTime = get_last_lap_time(p_plus2)
            p_plus3_LapTime = get_last_lap_time(p_plus3)

            def get_lap_start_time(driver_name):

                if driver_name is None:
                    return pd.NaT

                driver_lap = data[
                    (data["Driver"] == driver_name) &
                    (data["LapNumber"] == lap)
                ]

                if driver_lap.empty:
                    return pd.NaT

                return driver_lap["LapStartTime"].iloc[0]


            p_minus1_lap_start = get_lap_start_time(p_minus1_driver)
            p_minus2_lap_start = get_lap_start_time(p_minus2_driver)
            p_minus3_lap_start = get_lap_start_time(p_minus3_driver)

            p_plus1_lap_start = get_lap_start_time(p_plus1_driver)
            p_plus2_lap_start = get_lap_start_time(p_plus2_driver)
            p_plus3_lap_start = get_lap_start_time(p_plus3_driver)


            # Time separation from current driver
            p_minus1_time_delta = (
                current_lap_start - p_minus1_lap_start
                if pd.notna(p_minus1_lap_start) and pd.notna(current_lap_start)
                else np.nan
            )

            p_minus2_time_delta = (
                current_lap_start - p_minus2_lap_start
                if pd.notna(p_minus2_lap_start) and pd.notna(current_lap_start)
                else np.nan
            )

            p_minus3_time_delta = (
                current_lap_start - p_minus3_lap_start
                if pd.notna(p_minus3_lap_start) and pd.notna(current_lap_start)
                else np.nan
            )

            p_plus1_time_delta = (
                p_plus1_lap_start - current_lap_start
                if pd.notna(p_plus1_lap_start) and pd.notna(current_lap_start)
                else np.nan
            )

            p_plus2_time_delta = (
                p_plus2_lap_start - current_lap_start
                if pd.notna(p_plus2_lap_start) and pd.notna(current_lap_start)
                else np.nan
            )

            p_plus3_time_delta = (
                p_plus3_lap_start - current_lap_start
                if pd.notna(p_plus3_lap_start) and pd.notna(current_lap_start)
                else np.nan
            )

            future = data[
                (data["LapNumber"] == lap + 5) &
                (data["Driver"] == driver)
            ]

            if future.empty:
                continue

            future_position = future["Position"].iloc[0]

            if pd.isna(future_position):
                continue

            target = Position - int(future_position)


            row = {
                "Driver": Driver,
                "Position": Position,
                "Time": Time,
                "Team": Team,
                "Compound": Compound,
                "TyreLife": TyreLife,
                "LapTime": LapTime,

                "p_minus1_time_delta": p_minus1_time_delta,
                "p_minus2_time_delta": p_minus2_time_delta,
                "p_minus3_time_delta": p_minus3_time_delta,

                "p_plus1_time_delta": p_plus1_time_delta,
                "p_plus2_time_delta": p_plus2_time_delta,
                "p_plus3_time_delta": p_plus3_time_delta,

                "p_minus1_mean_LapTime": p_minus1_mean_LapTime,
                "p_minus2_mean_LapTime": p_minus2_mean_LapTime,
                "p_minus3_mean_LapTime": p_minus3_mean_LapTime,

                "p_minus1_median_LapTime": p_minus1_median_LapTime,
                "p_minus2_median_LapTime": p_minus2_median_LapTime,
                "p_minus3_median_LapTime": p_minus3_median_LapTime,

                "p_plus1_mean_LapTime": p_plus1_mean_LapTime,
                "p_plus2_mean_LapTime": p_plus2_mean_LapTime,
                "p_plus3_mean_LapTime": p_plus3_mean_LapTime,

                "p_plus1_median_LapTime": p_plus1_median_LapTime,
                "p_plus2_median_LapTime": p_plus2_median_LapTime,
                "p_plus3_median_LapTime": p_plus3_median_LapTime,

                "p_3_minus1_mean_LapTime": p_3_minus1_mean_LapTime,
                "p_3_minus2_mean_LapTime": p_3_minus2_mean_LapTime,
                "p_3_minus3_mean_LapTime": p_3_minus3_mean_LapTime,

                "p_3_minus1_median_LapTime": p_3_minus1_median_LapTime,
                "p_3_minus2_median_LapTime": p_3_minus2_median_LapTime,
                "p_3_minus3_median_LapTime": p_3_minus3_median_LapTime,

                "p_3_plus1_mean_LapTime": p_3_plus1_mean_LapTime,
                "p_3_plus2_mean_LapTime": p_3_plus2_mean_LapTime,
                "p_3_plus3_mean_LapTime": p_3_plus3_mean_LapTime,

                "p_3_plus1_median_LapTime": p_3_plus1_median_LapTime,
                "p_3_plus2_median_LapTime": p_3_plus2_median_LapTime,
                "p_3_plus3_median_LapTime": p_3_plus3_median_LapTime,

                "p_minus1_LapTime": p_minus1_LapTime,
                "p_minus2_LapTime": p_minus2_LapTime,
                "p_minus3_LapTime": p_minus3_LapTime,

                "p_plus1_LapTime": p_plus1_LapTime,
                "p_plus2_LapTime": p_plus2_LapTime,
                "p_plus3_LapTime": p_plus3_LapTime,

                "target": target
            }

            rows.append(row)

    return pd.DataFrame(rows)
```

```python
datasets = []

for i in df["Season"].unique():
    for j in df.loc[df["Season"] == i, "Round"].unique():

        sdf = df[
            (df["Season"] == i) &
            (df["Round"] == j)
        ].copy()

        datasets.append(create_data_for_round(sdf))

dataset = pd.concat(datasets, ignore_index=True)
```
