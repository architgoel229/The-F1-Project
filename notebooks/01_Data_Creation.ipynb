```python
import numpy as np
import pandas as pd
import fastf1 as f1
from data_profiling import ProfileReport
import seaborn as sns
import matplotlib.pyplot as plt
import os
```

```python
sessions = dict()
for i in range(2020, 2026):
    races = list()
    j=0
    while True:
        j+=1
        try:
            races.append(f1.get_session(i, j, 5))
        except ValueError:
            print(i, "done")
            break
    sessions[i]=races.copy()
```

# Load

```python

for key, value in sessions.items():
    for k in value:
        k.load(laps=True, telemetry=False, weather=False, messages=False)
        print("*"*20)
        print(f"{key}, {k} done")
        print("*"*20)

    print("*"*20)
    print(f"{key} done")
    print("*"*20)
```

# dataset altering

```python
highErr=list()
flag =0
for key, value in sessions.items():
    
    i=1
    for k in value:
        print("_"*20)
        print(f"{key}, {k}")
        print("_"*20)
        print(k.laps["IsAccurate"].mean(), end="\n\n")
        if k.laps["IsAccurate"].mean()<0.75:
            highErr.append(k)

print(flag)
print(highErr, len(highErr))
```
