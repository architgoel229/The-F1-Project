```python
import numpy as np
import pandas as pd
import os
```

```python
rounds = os.listdir(f"../data/2026/")
races=list()
for j in rounds:
    races.append(pd.read_csv(f"../data/2026/{j}/laps.csv"))
df = races
```

```python
n = len(df)
i = 0
races=list()
while i<n:
    races.append(pd.concat([df[i],pd.DataFrame({"Round": [i+1 for _ in range(df[i].shape[0])]})], axis=1))
    i+=1
season = pd.concat(races, axis=0)
season["Season"]=int(2026)
data=season
```

```python
data=data[(data["PitInTime"].isnull())&(data["PitOutTime"].isnull())].copy()
```

```python
from datetime import datetime
```

```python
def to_seconds(x):
    return x.total_seconds()
```

```python
colList = data.columns
for item in colList:
    if "time" in item.lower():
        data[item] = pd.to_timedelta(data[item])
        data[item] = data[item].apply(to_seconds)
```

```python
missing_tyreLife = data[data["TyreLife"].isna()].copy().sort_values(["Season", "Round", "LapNumber"])
missing_tyreLife[
    [
        "Time",
        "Driver",
        "DriverNumber",
        "LapNumber",
        "Position",
        "Stint",
        "PitOutTime",
        "PitInTime",
        "TrackStatus",
        "Deleted",
        "DeletedReason",
        "FastF1Generated",
        "IsAccurate", "Season", "Round"
    ]
].sort_values(["Season", "Round", "LapNumber"])
```

```python
for i in range(missing_tyreLife.shape[0]):

    row = missing_tyreLife.iloc[i]

    lap = row["LapNumber"]

    if lap == 1:
        continue

    driver = row["Driver"]
    season = row["Season"]
    round_ = row["Round"]

    current_condition = (
        (data["LapNumber"] == lap) &
        (data["Driver"] == driver) &
        (data["Season"] == season) &
        (data["Round"] == round_)
    )

    previous_condition = (
        (data["LapNumber"] == lap - 1) &
        (data["Driver"] == driver) &
        (data["Season"] == season) &
        (data["Round"] == round_)
    )
    next_condition = (
            (data["LapNumber"] == lap + 1) &
            (data["Driver"] == driver) &
            (data["Season"] == season) &
            (data["Round"] == round_)
        )


    previous = data.loc[previous_condition]
    Next = data.loc[next_condition]


    if previous.empty:
        continue

    previous_tyre_life = (
        previous["TyreLife"].iloc[0]
        if not previous.empty
        else np.nan
    )

    next_tyre_life = (
        Next["TyreLife"].iloc[0]
        if not Next.empty
        else np.nan
    )


    if pd.notna(previous_tyre_life) and pd.notna(next_tyre_life):

        previous_estimate = previous_tyre_life + 1
        next_estimate = next_tyre_life - 1

        if previous_estimate == next_estimate:
            data.loc[current_condition, "TyreLife"] = previous_estimate
            print("edited")

        else:
            data.loc[current_condition, "TyreLife"] = next_estimate
            print("edited")

    elif pd.notna(previous_tyre_life):

        data.loc[current_condition, "TyreLife"] = previous_tyre_life + 1
        print("edited")

    elif pd.notna(next_tyre_life):

        data.loc[current_condition, "TyreLife"] = next_tyre_life - 1
        print("edited")
```
