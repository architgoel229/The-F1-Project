```python
import numpy as np
import pandas as pd
import fastf1 as f1
from data_profiling import ProfileReport
import seaborn as sns
import matplotlib.pyplot as plt
import os
```

```python
races = list()
j=0
while True:
    j+=1
    try:
        races.append(f1.get_session(2026, j, 5))
    except ValueError:
        print(2026, "done")
        break
```

```python
for k in races:
    k.load(laps=True, telemetry=False, weather=False, messages=False)
    print("*"*20)
    print(f" {k} done")
    print("*"*20)
```

```python
n = len(races)
i=0
while i<n:
    os.makedirs(f"../data/2026/{i+1}/",  exist_ok=True)
    try: 
        races[i].laps.to_csv(f"../data/2026/{i+1}/laps.csv", index=False)
        i+=1
    except:
        break
```
